```python
import numpy as np
import scipy as sp
```

```python
g = 2

def int1(x):
    return gaussian(x) * (x**4) * g * (1/np.sqrt(np.pi))

def int2(x):
    return sp.special.eval_hermite(2, x) * gaussian(x) * (x**4) * g * (1/np.sqrt(np.pi)) * (1/np.sqrt(8))
    
def int3(x):
    return sp.special.eval_hermite(4, x) * gaussian(x) * (x**4) * g * (1/np.sqrt(np.pi)) * (1/np.sqrt(384))

def int4(x):
    return sp.special.eval_hermite(6, x) * gaussian(x) * (x**4) * g * (1/np.sqrt(np.pi)) * (1/np.sqrt(46080))

def gaussian(x):
    return np.exp(-(x**2))
```

```python
samples, weights = np.polynomial.legendre.leggauss(99)

b = 10.0
a = -10.0

c1 = (b-a)/2
c2 = (b+a)/2
```

```python
integral1 = c1 * sum([weights[i] * int1(c1*samples[i] + c2) for i in range(len(weights))])

print(integral1)
```

```python
integral2 = c1 * sum([weights[i] * int2(c1*samples[i] + c2) for i in range(len(weights))])

print(integral2)
```

```python
integral3 = c1 * sum([weights[i] * int3(c1*samples[i] + c2) for i in range(len(weights))])

print(integral3)
```
